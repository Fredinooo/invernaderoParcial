```python
# Constantes
tempOptMin = 18
tempOptMax = 26
humOptMin = 60
humOptMax = 80

tempCritMin = 12
tempCritMax = 32
humCritMin = 40
humCritMax = 90

tempValidaMin = -10
tempValidaMax = 60
humValidaMin = 0
humValidaMax = 100

maxLecturas = 8


def formatear(valor):
    if valor == int(valor):
        return str(int(valor))
    return str(valor).replace(".", ",")


def formatearDecimales(valor):
    return "{:.2f}".format(valor).replace(".", ",")


# Variables
numLectura = 0
temp = 0.0
hum = 0.0
estado = ""
accion = ""
resultado = ""
terminar = False

minTemp = 0.0
maxTemp = 0.0
minHum = 0.0
maxHum = 0.0
sumTemp = 0.0
sumHum = 0.0

secTemp = []
secHum = []

print("INVERNADERO EN OBSERVACION")

while not terminar:
    # 1. Pedir y validar la lectura
    valido = False
    while not valido:
        temp = float(input("Ingrese la temperatura: ").replace(",", "."))
        hum = float(input("Ingrese la humedad: ").replace(",", "."))
        if temp < tempValidaMin or temp > tempValidaMax or hum < humValidaMin or hum > humValidaMax:
            print("Dato inválido. Ingrese nuevamente la lectura (no se cuenta).")
        else:
            valido = True

    numLectura += 1
    secTemp.append(temp)
    secHum.append(hum)
    sumTemp += temp
    sumHum += hum

    # Mínimos y máximos
    if numLectura == 1:
        minTemp = temp
        maxTemp = temp
        minHum = hum
        maxHum = hum
    else:
        if temp < minTemp:
            minTemp = temp
        if temp > maxTemp:
            maxTemp = temp
        if hum < minHum:
            minHum = hum
        if hum > maxHum:
            maxHum = hum

    # 2 y 3. Evaluar el estado y las condiciones
    # Condición A: Valor crítico
    if temp < tempCritMin or temp > tempCritMax or hum < humCritMin or hum > humCritMax:
        estado = "Crítico"
        accion = "Alerta crítica: activar protocolo de emergencia"
        resultado = "Alerta crítica"
        terminar = True

    # Condición B: Óptimo
    elif tempOptMin <= temp <= tempOptMax and humOptMin <= hum <= humOptMax:
        estado = "Óptimo"
        terminar = True
        if numLectura == 1:
            accion = "Cultivo en condiciones óptimas"
            resultado = "Condiciones óptimas"
        else:
            accion = "Cultivo estabilizado"
            resultado = "Cultivo estabilizado"

    # Condición C: Observación
    else:
        estado = "Observación"

        if numLectura == maxLecturas:
            accion = "Revisión técnica de los equipos"
            resultado = "Revisión técnica"
            terminar = True
        else:
            acciones_list = []
            if temp > tempOptMax:
                acciones_list.append("Ventilar")
            elif temp < tempOptMin:
                acciones_list.append("Encender calefacción")

            if hum > humOptMax:
                acciones_list.append("deshumidificar")
            elif hum < humOptMin:
                acciones_list.append("nebulizar")

            accion = " y ".join(acciones_list)

    print("Lectura", numLectura, "|", formatear(temp), "|", formatear(hum), "|", estado, "|", accion)

# Cálculos finales
promTemp = sumTemp / numLectura
promHum = sumHum / numLectura

textoTemp = "; ".join(formatear(x) for x in secTemp)
textoHum = "; ".join(formatear(x) for x in secHum)

print()
print(f"Resultado: {resultado}. Lecturas tomadas: {numLectura}.")
print(f"Temperatura: mínima {formatear(minTemp)} – máxima {formatear(maxTemp)} – promedio {formatearDecimales(promTemp)}.")
print(f"Humedad: mínima {formatear(minHum)} – máxima {formatear(maxHum)} – promedio {formatearDecimales(promHum)}.")
print(f"Secuencia de temperaturas: {textoTemp}. Secuencia de humedades: {textoHum}.")
```
